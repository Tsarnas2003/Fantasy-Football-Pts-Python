```python
import pandas as pd
df = pd.read_csv("scrimmage.csv",header = None)
df.head()
```

```python
column_names_df = df.head(2)
column_names_df
```

```python
column_names = []
for column in column_names_df.columns:
    column_values = column_names_df[column].astype(str)
    column_name = ' '.join(column_values)
    column_name = column_name.replace('nan', '')
    column_names.append(column_name)
column_names

```

```python
player_data_df = df.tail(len(df)-len(column_names_df))
player_data_df = player_data_df.drop(623)
player_data_df
```

```python
player_data_df.columns = column_names



player_data_df.head()
```

```python
column_names
keep_cols =[
 ' Player',' Age',' Team',' Pos',' G','Receiving Tgt','Receiving Rec','Receiving Yds',
 'Receiving Y/R','Receiving TD','Rushing Att','Rushing Yds','Rushing TD','Scrimmage Touch','Scrimmage YScm',
 'Scrimmage RRTD']
player_data_df = player_data_df[keep_cols]
player_data_df.head()
```

```python
pd.set_option('display.max_rows',100)
player_data_df.head(100)
```

```python
vc =player_data_df[' Player'].value_counts()
multi_team_player = vc[vc > 1].index
multi_team_player_df = player_data_df[player_data_df[' Player'].isin(multi_team_player)]
multi_team_player_df=multi_team_player_df[multi_team_player_df[' Team'].str.contains('TM')]
```

```python
single_team_player_df = player_data_df[~player_data_df[' Player'].isin(multi_team_player)]
player_dfs = [single_team_player_df,multi_team_player_df]
player_df = pd.concat(player_dfs)
player_df.head(100)
```

```python
player_df[' Pos'].value_counts()
keep_QB = ['QB']
qb_scrimmage_df = player_df[player_df[' Pos'].isin(keep_QB)].copy()
keep_pos = ['WR','RB','TE']
player_df = player_df[player_df[' Pos'].isin(keep_pos)]
qb_scrimmage_df
```

```python
player_df['Pts'] = player_df['Scrimmage YScm'].astype(float) / 10 + player_df['Scrimmage RRTD'].astype(float) * 6 + player_df['Receiving Rec'].astype(int)*1
```

```python
player_df.sort_values('Pts',ascending=False)
```

```python
dfqb = pd.read_csv("PassingYards2024.csv")
pd.set_option('display.max_rows',100)
dfqb = dfqb.drop(111)
dfqb.head(100)
```

```python
single_team_qb_df = dfqb[~dfqb['Player'].isin(multi_team_qb)]
qb_dfs = [single_team_qb_df,multi_team_qb_df]
qb_df = pd.concat(qb_dfs)
```

```python
dfqb.columns
keep_cols1 =['Player', 'Age', 'Team', 'Pos', 'G', 'Cmp', 'Att', 'Yds', 'TD', 'Int']
qb_df = qb_df[keep_cols1]    
qb_df = qb_df[qb_df['Pos'] == 'QB']
qb_df
```

```python
qb_df.columns = qb_df.columns.str.strip()
qb_scrimmage_df.columns = qb_scrimmage_df.columns.str.strip()
qb_scrimmage_correct = qb_scrimmage_df[['Player', 'Scrimmage YScm', 'Scrimmage RRTD']].copy()
qb_df = qb_df.merge(qb_scrimmage_correct, on='Player', how='left')


#qb_df['Pts'] = qb_df['Yds'].astype(float) / 25 + qb_df['TD'].astype(float) * 4 - qb_df['Int'].astype(float) * 2 + qb_df['Scrimmage YScm'].astype(float) / 10 + qb_df['Scrimmage RRTD'].astype(float) * 6
qb_df.columns
```

```python

keep_cols3= ['Player', 'Age', 'Team', 'Pos', 'G', 'Cmp', 'Att', 'Yds', 'TD', 'Int',
        'Scrimmage YScm', 'Scrimmage RRTD']
qb_df = qb_df[keep_cols3]
qb_df.head(50)
```

```python
qb_df['Pts'] = qb_df['Yds'].astype(float) / 25 + qb_df['TD'].astype(float) * 4 - qb_df['Int'].astype(float) * 2 + qb_df['Scrimmage YScm'].astype(float) / 10 + qb_df['Scrimmage RRTD'].astype(float) * 6
qb_df.sort_values('Pts',ascending=False)
```

```python
# Combine QB and player fantasy points
player_df.columns = player_df.columns.str.strip()
qb_pts_df = qb_df[['Player','Age','Team','Pos','G','Pts']]
player_pts_df = player_df[['Player','Age','Team','Pos','G','Pts']]

final_df = pd.concat([qb_pts_df, player_pts_df], ignore_index=True)

# Replace NaN fantasy points with 0
final_df['Pts'] = final_df['Pts'].fillna(0)
final_df['Age']= final_df['Age'].astype(int)
final_df['G']= final_df['G'].astype(int)

# Sort by fantasy points (highest first) and reset index
final_df = final_df.sort_values('Pts', ascending=False).reset_index(drop=True)

print(final_df.head(100))  # Top 100 players
```
